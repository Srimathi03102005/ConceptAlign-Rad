```python
REPORT_FILE = "/kaggle/input/datasets/xyz2005/notebook-7/generated_reports.csv"

df = pd.read_csv(REPORT_FILE)

print("Number of reports:", len(df))
print("Columns:")
print(df.columns.tolist())

display(df.head(3))
```

```python
print(df.columns.tolist())
```

```python
print("Available columns:")
for i, col in enumerate(df.columns):
    print(i, "->", col)
```

```python
display(df.iloc[:2])
```

```python
GENERATED_COL = "generated_report"
REFERENCE_COL = "report"
```

```python
generated_reports = (
    df[GENERATED_COL]
    .fillna("")
    .astype(str)
    .tolist()
)

reference_reports = (
    df[REFERENCE_COL]
    .fillna("")
    .astype(str)
    .tolist()
)

print("Generated reports:", len(generated_reports))
print("Reference reports:", len(reference_reports))
```

```python
assert len(generated_reports) == len(reference_reports)

print("✓ Number of reports:", len(generated_reports))

print("\n--- GENERATED ---")
print(generated_reports[0])

print("\n--- REFERENCE ---")
print(reference_reports[0])
```

```python
from f1chexbert import F1CheXbert

print("Loading CheXbert...")

chexbert = F1CheXbert()

print("✓ CheXbert loaded")
```

```python
print("Calculating F1CheXbert...")
print("This may take a few minutes for 496 reports.")

chexbert_result = chexbert(
    hyps=generated_reports,
    refs=reference_reports
)

print("✓ CheXbert evaluation completed")
```

```python
print(type(chexbert_result))
print(chexbert_result)
```

```python
accuracy = chexbert_result[0]
accuracy_not_averaged = chexbert_result[1]
class_report = chexbert_result[2]
class_report_5 = chexbert_result[3]

print("CheXbert Accuracy:", accuracy)

print("\nCheXbert class report:")
print(class_report)
```

```python
chexbert_score = accuracy

print("\n================================")
print("F1CheXbert Score:", chexbert_score)
print("================================")
```

```python
from radgraph import F1RadGraph

print("✓ RadGraph imported")
```

```python
print("Loading RadGraph...")

f1radgraph = F1RadGraph(
    reward_level="all",
    model_type="radgraph-xl"
)

print("✓ RadGraph loaded")
```

```python
print("Calculating RadGraph...")
print("Processing", len(generated_reports), "reports...")

radgraph_result = f1radgraph(
    hyps=generated_reports,
    refs=reference_reports
)

print("✓ RadGraph evaluation completed")
```

```python
mean_reward = radgraph_result[0]

print("RadGraph mean reward:")
print(mean_reward)
```

```python
rg_e, rg_er, rg_combined = mean_reward

print("======================================")
print("RADGRAPH RESULTS")
print("======================================")

print("Entity F1:                 ", rg_e)
print("Entity + Relation F1:      ", rg_er)
print("Combined RadGraph Score:   ", rg_combined)
```

```python
clinical_results = {
    "F1CheXbert": float(chexbert_score),
    "RadGraph_Entity_F1": float(rg_e),
    "RadGraph_Entity_Relation_F1": float(rg_er),
    "RadGraph_Combined": float(rg_combined)
}

clinical_results
```
