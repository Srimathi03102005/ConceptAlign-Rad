```python
import pandas as pd
import numpy as np
import re
```

```python
train_df = pd.read_csv("/kaggle/input/datasets/xyz2005/notebook-3/train_reports.csv")
valid_df = pd.read_csv("/kaggle/input/datasets/xyz2005/notebook-3/valid_reports.csv")
test_df = pd.read_csv("/kaggle/input/datasets/xyz2005/notebook-3/test_reports.csv")

print(train_df.head())
```

```python
DISEASES = {

    "Pneumonia":[
        "pneumonia"
    ],

    "Cardiomegaly":[
        "cardiomegaly",
        "enlarged heart"
    ],

    "Pleural Effusion":[
        "pleural effusion",
        "effusion"
    ],

    "Atelectasis":[
        "atelectasis"
    ],

    "Edema":[
        "edema",
        "pulmonary edema"
    ],

    "Pneumothorax":[
        "pneumothorax"
    ],

    "Consolidation":[
        "consolidation"
    ],

    "Lung Opacity":[
        "opacity",
        "lung opacity"
    ],

    "Nodule":[
        "nodule"
    ],

    "Mass":[
        "mass"
    ]
}
```

```python
def create_labels(report):

    report = str(report).lower()

    labels = {}

    for disease, keywords in DISEASES.items():

        found = 0

        for word in keywords:

            if word in report:

                found = 1
                break

        labels[disease] = found

    return pd.Series(labels)
```

```python
train_labels = train_df["report"].apply(create_labels)

train_concepts = pd.concat(
    [train_df, train_labels],
    axis=1
)

train_concepts.head()
```

```python
valid_labels = valid_df["report"].apply(create_labels)

valid_concepts = pd.concat(
    [valid_df, valid_labels],
    axis=1
)
```

```python
test_labels = test_df["report"].apply(create_labels)

test_concepts = pd.concat(
    [test_df, test_labels],
    axis=1
)
```

```python
for disease in DISEASES.keys():

    print(disease)

    print(train_concepts[disease].value_counts())

    print()
```

```python
train_concepts.to_csv(
    "/kaggle/working/train_concepts.csv",
    index=False
)

valid_concepts.to_csv(
    "/kaggle/working/valid_concepts.csv",
    index=False
)

test_concepts.to_csv(
    "/kaggle/working/test_concepts.csv",
    index=False
)
```
